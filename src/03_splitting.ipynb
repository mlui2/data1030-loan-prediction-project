This notebook loads the cleaned dataset and performs a chronological split
into training (2016–2017) and testing (2018 only)

```python
import pandas as pd

df = pd.read_pickle("../data/clean_data.pkl")
print("Loaded cleaned data:", df.shape)
```

```python
#Only 2018 loans in test
test_mask = df["issue_year"] == 2018

#Drop issue_year for no leakage
df = df.drop(columns="issue_year").drop(columns="issue_d")

test_df = df[test_mask]
train_df = df[~test_mask]
```

```python
train_df.to_pickle("../data/modeling/train_df.pkl")
test_df.to_pickle("../data/modeling/test_df.pkl")

print("Saved train/test splits:")
print("train_df.pkl →", train_df.shape)
print("test_df.pkl  →", test_df.shape)
```
